```python
#!/usr/bin/env python3
"""
Advanced Reproducible Analysis Script for NFS Ransomware Detector Overhead Evaluation

This script analyzes results from run_overhead_evaluation.sh with advanced statistical
methods and allows full reproducibility by re-parsing raw data (FIO JSON, pidstat logs).

NEW FEATURES:
  • Statistical Significance Testing (t-tests with p-values)
  • 95% Confidence Intervals (CI) for all metrics
  • Outlier Detection (IQR method) per measurement set
  • IOPS & Latency Analysis with significance testing
  • Trend Analysis (thermal degradation detection)
  • Normalized CPU Overhead (% of total CPU budget)
  • Effect Sizes (Cohen's d) for practical significance

Usage:
    python3 analyze_overhead_results.py <results.csv>
    python3 analyze_overhead_results.py <raw_data_directory>
    python3 analyze_overhead_results.py <results.csv> --reparse
    python3 analyze_overhead_results.py <results.csv> --detailed
    python3 analyze_overhead_results.py <raw_data_directory> --export-csv output.csv

Options:
    --reparse       Re-parse raw data (FIO JSON + pidstat logs) instead of using CSV
    --detailed      Show detailed per-run statistics
    --export-plots  Generate matplotlib plots (requires matplotlib)
    --export-csv    Export parsed data to CSV file (when using raw_data directory)
    --help          Show this help message

Output Interpretation:
  • p-value < 0.05:      Statistically significant difference
  • Effect size (d):     0.2=small, 0.5=medium, 0.8=large
  • Outliers:            Detected using IQR method (1.5×IQR bounds)
  • CI Margin:           95% confidence interval half-width
  • Trend (R²):          0.3+= notable trend (degradation or stabilization)
"""

import csv
import json
import sys
import os
import statistics
import re
import math
from pathlib import Path
from typing import Dict, List, Tuple, Optional
from scipy import stats


# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def parse_pidstat_log(log_path: str) -> Tuple[float, int]:
    """
    Parse pidstat log file and extract average CPU percentage.
    
    Returns:
        (avg_cpu_percent, sample_count)
    """
    if not os.path.exists(log_path):
        return (0.0, 0)
    
    cpu_values = []
    
    with open(log_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) < 7:
                continue
            
            # Find %CPU column (usually 3rd from end)
            try:
                cpu_val = float(parts[-3])
                if 0 <= cpu_val <= 100:
                    cpu_values.append(cpu_val)
            except (ValueError, IndexError):
                continue
    
    if not cpu_values:
        return (0.0, 0)
    
    return (statistics.mean(cpu_values), len(cpu_values))


def parse_fio_json(json_path: str) -> Dict:
    """
    Parse FIO JSON output and extract relevant metrics.
    
    Returns:
        dict with keys: runtime_sec, bw_kbps, iops, lat_ns
    """
    if not os.path.exists(json_path):
        return {}
    
    with open(json_path, 'r') as f:
        data = json.load(f)
    
    job = data.get('jobs', [{}])[0]
    
    # Determine if read or write workload
    read_data = job.get('read', {})
    write_data = job.get('write', {})
    
    # Use whichever has more data
    if read_data.get('io_bytes', 0) > write_data.get('io_bytes', 0):
        io_data = read_data
        io_type = 'read'
    else:
        io_data = write_data
        io_type = 'write'
    
    return {
        'runtime_sec': job.get('job_runtime', 0) / 1000.0,
        'bw_kbps': io_data.get('bw', 0),
        'iops': io_data.get('iops', 0),
        'lat_ns': io_data.get('lat_ns', {}).get('mean', 0),
        'io_type': io_type,
        'total_io_mb': io_data.get('io_bytes', 0) / (1024 * 1024)
    }


def reparse_raw_data(raw_data_dir: str) -> List[Dict]:
    """
    Re-parse all raw data (FIO JSON + pidstat logs) to reconstruct results.
    This allows full reproducibility independent of the original script's parsing.
    
    Args:
        raw_data_dir: Path to raw_data directory (or CSV path, will auto-detect)
    """
    # If CSV path given, construct raw_data_dir path
    if raw_data_dir.endswith('.csv'):
        csv_dir = os.path.dirname(raw_data_dir)
        csv_basename = os.path.splitext(os.path.basename(raw_data_dir))[0]
        raw_data_dir = os.path.join(csv_dir, f"raw_data_{csv_basename}")
    
    if not os.path.exists(raw_data_dir):
        print(f"Error: Raw data directory not found: {raw_data_dir}")
        sys.exit(1)
    
    if not os.path.isdir(raw_data_dir):
        print(f"Error: Not a directory: {raw_data_dir}")
        sys.exit(1)
    
    print(f"Re-parsing raw data from: {raw_data_dir}")
    
    results = []
    
    # Find all FIO JSON files
    fio_files = sorted(Path(raw_data_dir).glob("fio_*.json"))
    
    if not fio_files:
        print(f"Error: No FIO JSON files found in {raw_data_dir}")
        print(f"Expected files like: fio_baseline_run1.json, fio_detector_run1.json")
        sys.exit(1)
    
    print(f"Found {len(fio_files)} FIO result files")
    
    for fio_file in fio_files:
        # Extract mode and run_num from filename: fio_baseline_run1.json
        match = re.match(r'fio_(baseline|detector)_run(\d+)\.json', fio_file.name)
        if not match:
            print(f"Warning: Skipping unrecognized file: {fio_file.name}")
            continue
        
        mode = match.group(1)
        run_num = int(match.group(2))
        
        # Parse FIO data
        fio_data = parse_fio_json(str(fio_file))
        
        # Parse nfsd pidstat log
        nfsd_log = os.path.join(raw_data_dir, f"pidstat_nfsd_{mode}_run{run_num}.log")
        nfsd_cpu, nfsd_samples = parse_pidstat_log(nfsd_log)
        
        # Parse detector pidstat log (only for detector mode)
        detector_cpu = 0.0
        detector_samples = 0
        if mode == 'detector':
            detector_log = os.path.join(raw_data_dir, f"pidstat_detector_run{run_num}.log")
            detector_cpu, detector_samples = parse_pidstat_log(detector_log)
        
        results.append({
            'mode': mode,
            'run_num': run_num,
            'fio_runtime_sec': fio_data.get('runtime_sec', 0),
            'fio_bw_kbps': fio_data.get('bw_kbps', 0),
            'fio_iops': fio_data.get('iops', 0),
            'fio_lat_ns': fio_data.get('lat_ns', 0),
            'nfsd_avg_cpu_pct': nfsd_cpu,
            'nfsd_samples': nfsd_samples,
            'detector_avg_cpu_pct': detector_cpu,
            'detector_samples': detector_samples,
            'raw_source': 'reparsed'
        })
    
    # Sort by mode and run_num
    results.sort(key=lambda x: (x['mode'], x['run_num']))
    
    print(f"Successfully parsed {len(results)} runs")
    print(f"  - Baseline: {len([r for r in results if r['mode'] == 'baseline'])} runs")
    print(f"  - Detector: {len([r for r in results if r['mode'] == 'detector'])} runs")
    
    return results


def load_csv_results(csv_path: str) -> List[Dict]:
    """Load results from CSV file."""
    results = []
    with open(csv_path, 'r') as f:
        reader = csv.DictReader(f)
        for row in reader:
            # Convert numeric fields
            row['run_num'] = int(row['run_num'])
            row['fio_runtime_sec'] = float(row['fio_runtime_sec'])
            row['fio_bw_kbps'] = float(row['fio_bw_kbps'])
            row['nfsd_avg_cpu_pct'] = float(row['nfsd_avg_cpu_pct'])
            row['detector_avg_cpu_pct'] = float(row['detector_avg_cpu_pct'])
            row['raw_source'] = 'csv'
            results.append(row)
    return results


# ============================================================================
# ANALYSIS FUNCTIONS
# ============================================================================

def calculate_statistics(values: List[float]) -> Dict:
    """Calculate statistical measures for a list of values."""
    if not values:
        return {
            'mean': 0, 'median': 0, 'std': 0,
            'min': 0, 'max': 0, 'p25': 0, 'p75': 0,
            'ci_lower': 0, 'ci_upper': 0, 'ci_margin': 0
        }
    
    mean = statistics.mean(values)
    std = statistics.stdev(values) if len(values) > 1 else 0
    
    # 95% Confidence Interval usando t-distribution
    se = std / math.sqrt(len(values)) if std > 0 else 0
    t_val = stats.t.ppf(0.975, len(values) - 1) if len(values) > 1 else 1.96
    ci_margin = t_val * se
    
    return {
        'mean': mean,
        'median': statistics.median(values),
        'std': std,
        'min': min(values),
        'max': max(values),
        'p25': statistics.quantiles(values, n=4)[0] if len(values) >= 4 else min(values),
        'p75': statistics.quantiles(values, n=4)[2] if len(values) >= 4 else max(values),
        'ci_lower': mean - ci_margin,
        'ci_upper': mean + ci_margin,
        'ci_margin': ci_margin,
        'count': len(values)
    }


def detect_outliers(values: List[Tuple[int, float]]) -> Tuple[List[int], List[float]]:
    """
    Detect outliers using IQR method.
    
    Args:
        values: List of (run_num, value) tuples
    
    Returns:
        (outlier_run_nums, outlier_values)
    """
    if len(values) < 4:
        return ([], [])
    
    just_values = [v[1] for v in values]
    q1 = statistics.quantiles(just_values, n=4)[0]
    q3 = statistics.quantiles(just_values, n=4)[2]
    iqr = q3 - q1
    
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    
    outliers_runs = []
    outliers_vals = []
    
    for run_num, val in values:
        if val < lower_bound or val > upper_bound:
            outliers_runs.append(run_num)
            outliers_vals.append(val)
    
    return (outliers_runs, outliers_vals)


def perform_t_test(baseline: List[float], detector: List[float]) -> Dict:
    """
    Perform independent samples t-test between baseline and detector.
    
    Returns:
        dict with: t_statistic, p_value, is_significant (p < 0.05)
    """
    if len(baseline) < 2 or len(detector) < 2:
        return {
            't_statistic': 0,
            'p_value': 1.0,
            'is_significant': False,
            'effect_size': 0
        }
    
    t_stat, p_val = stats.ttest_ind(baseline, detector)
    
    # Cohen's d effect size
    pooled_std = math.sqrt((statistics.variance(baseline) + 
                           statistics.variance(detector)) / 2)
    cohens_d = (statistics.mean(detector) - statistics.mean(baseline)) / pooled_std if pooled_std > 0 else 0
    
    return {
        't_statistic': t_stat,
        'p_value': p_val,
        'is_significant': p_val < 0.05,
        'effect_size': cohens_d
    }


def analyze_trends(results: List[Dict], metric_key: str, mode: str) -> Dict:
    """
    Analyze trends over runs (e.g., thermal degradation).
    
    Returns:
        dict with: slope, r_squared, is_degrading
    """
    mode_data = [r for r in results if r['mode'] == mode]
    
    if len(mode_data) < 3:
        return {'slope': 0, 'r_squared': 0, 'is_degrading': False}
    
    x = [r['run_num'] for r in mode_data]
    y = [r[metric_key] for r in mode_data]
    
    # Linear regression
    x_mean = statistics.mean(x)
    y_mean = statistics.mean(y)
    
    numerator = sum((x[i] - x_mean) * (y[i] - y_mean) for i in range(len(x)))
    denominator = sum((x[i] - x_mean) ** 2 for i in range(len(x)))
    
    if denominator == 0:
        return {'slope': 0, 'r_squared': 0, 'is_degrading': False}
    
    slope = numerator / denominator
    intercept = y_mean - slope * x_mean
    
    # R-squared
    ss_res = sum((y[i] - (slope * x[i] + intercept)) ** 2 for i in range(len(x)))
    ss_tot = sum((y[i] - y_mean) ** 2 for i in range(len(y)))
    
    r_squared = 1 - (ss_res / ss_tot) if ss_tot > 0 else 0
    
    return {
        'slope': slope,
        'r_squared': r_squared,
        'is_degrading': slope > 0  # Runtime increases over time
    }


def calculate_normalized_overhead(detector_cpu: float, baseline_cpu: float, 
                                 total_cpu_budget: float = 800) -> float:
    """
    Calculate CPU overhead as percentage of total CPU budget.
    
    Args:
        detector_cpu: Average CPU % of detector process
        baseline_cpu: Average CPU % of nfsd in baseline
        total_cpu_budget: Total available CPU (default: 8 cores * 100%)
    
    Returns:
        Normalized overhead as % of total budget
    """
    delta = detector_cpu
    return (delta / total_cpu_budget) * 100


def analyze_results(results: List[Dict], detailed: bool = False) -> None:
    """Perform comprehensive analysis of overhead results."""
    
    # Split by mode
    baseline = [r for r in results if r['mode'] == 'baseline']
    detector = [r for r in results if r['mode'] == 'detector']
    
    if not baseline or not detector:
        print("Error: Missing baseline or detector runs")
        sys.exit(1)
    
    print("\n" + "=" * 80)
    print(" ADVANCED OVERHEAD EVALUATION ANALYSIS (with Statistical Significance)")
    print("=" * 80)
    print(f"\nDataset: {len(baseline)} baseline runs, {len(detector)} detector runs")
    
    # ========================================================================
    # RUNTIME ANALYSIS
    # ========================================================================
    baseline_times = [r['fio_runtime_sec'] for r in baseline]
    detector_times = [r['fio_runtime_sec'] for r in detector]
    
    baseline_stats = calculate_statistics(baseline_times)
    detector_stats = calculate_statistics(detector_times)
    
    time_overhead_pct = ((detector_stats['median'] - baseline_stats['median']) 
                         / baseline_stats['median'] * 100)
    
    # Outlier detection
    baseline_times_with_run = [(r['run_num'], r['fio_runtime_sec']) for r in baseline]
    detector_times_with_run = [(r['run_num'], r['fio_runtime_sec']) for r in detector]
    baseline_outliers, _ = detect_outliers(baseline_times_with_run)
    detector_outliers, _ = detect_outliers(detector_times_with_run)
    
    # T-test
    runtime_ttest = perform_t_test(baseline_times, detector_times)
    
    # Trend analysis
    baseline_trend = analyze_trends(results, 'fio_runtime_sec', 'baseline')
    detector_trend = analyze_trends(results, 'fio_runtime_sec', 'detector')
    
    print("\n" + "-" * 80)
    print("RUNTIME PERFORMANCE")
    print("-" * 80)
    print(f"Baseline:    {baseline_stats['median']:.3f}s ± {baseline_stats['ci_margin']:.3f}s (95% CI)")
    print(f"             {baseline_stats['std']:.3f}s std dev, range [{baseline_stats['min']:.3f}-{baseline_stats['max']:.3f}]s")
    if baseline_outliers:
        print(f"             ⚠ Outliers detected in runs: {baseline_outliers}")
    
    print(f"Detector:    {detector_stats['median']:.3f}s ± {detector_stats['ci_margin']:.3f}s (95% CI)")
    print(f"             {detector_stats['std']:.3f}s std dev, range [{detector_stats['min']:.3f}-{detector_stats['max']:.3f}]s")
    if detector_outliers:
        print(f"             ⚠ Outliers detected in runs: {detector_outliers}")
    
    print(f"\nOverhead:    {time_overhead_pct:+.2f}%  (p={runtime_ttest['p_value']:.4f}, "
          f"effect size={runtime_ttest['effect_size']:.3f})")
    if runtime_ttest['is_significant']:
        print(f"             ✓ STATISTICALLY SIGNIFICANT (p < 0.05)")
    else:
        print(f"             ✗ NOT statistically significant (p ≥ 0.05)")
    
    if detector_trend['r_squared'] > 0.3:
        print(f"             ⚠ Degradation trend detected (R²={detector_trend['r_squared']:.3f})")
    
    # ========================================================================
    # BANDWIDTH ANALYSIS
    # ========================================================================
    baseline_bw = [r['fio_bw_kbps'] for r in baseline]
    detector_bw = [r['fio_bw_kbps'] for r in detector]
    
    baseline_bw_stats = calculate_statistics(baseline_bw)
    detector_bw_stats = calculate_statistics(detector_bw)
    
    bw_overhead_pct = ((baseline_bw_stats['median'] - detector_bw_stats['median'])
                       / baseline_bw_stats['median'] * 100)
    
    bw_ttest = perform_t_test(baseline_bw, detector_bw)
    
    baseline_bw_outliers, _ = detect_outliers([(r['run_num'], r['fio_bw_kbps']) for r in baseline])
    detector_bw_outliers, _ = detect_outliers([(r['run_num'], r['fio_bw_kbps']) for r in detector])
    
    print("\n" + "-" * 80)
    print("BANDWIDTH PERFORMANCE")
    print("-" * 80)
    print(f"Baseline:    {baseline_bw_stats['median']:.0f} ± {baseline_bw_stats['ci_margin']:.0f} KB/s (95% CI)")
    print(f"             {baseline_bw_stats['std']:.0f} KB/s std dev")
    if baseline_bw_outliers:
        print(f"             ⚠ Outliers in runs: {baseline_bw_outliers}")
    
    print(f"Detector:    {detector_bw_stats['median']:.0f} ± {detector_bw_stats['ci_margin']:.0f} KB/s (95% CI)")
    print(f"             {detector_bw_stats['std']:.0f} KB/s std dev")
    if detector_bw_outliers:
        print(f"             ⚠ Outliers in runs: {detector_bw_outliers}")
    
    print(f"\nReduction:   {bw_overhead_pct:+.2f}%  (p={bw_ttest['p_value']:.4f}, "
          f"effect size={bw_ttest['effect_size']:.3f})")
    if bw_ttest['is_significant']:
        print(f"             ✓ STATISTICALLY SIGNIFICANT (p < 0.05)")
    else:
        print(f"             ✗ NOT statistically significant")
    
    # ========================================================================
    # CPU ANALYSIS
    # ========================================================================
    baseline_cpu = [r['nfsd_avg_cpu_pct'] for r in baseline]
    detector_cpu_nfsd = [r['nfsd_avg_cpu_pct'] for r in detector]
    detector_cpu_det = [r['detector_avg_cpu_pct'] for r in detector]
    
    baseline_cpu_stats = calculate_statistics(baseline_cpu)
    detector_cpu_nfsd_stats = calculate_statistics(detector_cpu_nfsd)
    detector_cpu_det_stats = calculate_statistics(detector_cpu_det)
    
    cpu_nfsd_overhead_pct = detector_cpu_nfsd_stats['median'] - baseline_cpu_stats['median']
    cpu_ttest = perform_t_test(baseline_cpu, detector_cpu_nfsd)
    
    baseline_cpu_outliers, _ = detect_outliers([(r['run_num'], r['nfsd_avg_cpu_pct']) for r in baseline])
    detector_cpu_outliers, _ = detect_outliers([(r['run_num'], r['nfsd_avg_cpu_pct']) for r in detector])
    
    normalized_overhead = calculate_normalized_overhead(
        detector_cpu_det_stats['median'],
        baseline_cpu_stats['median']
    )
    
    print("\n" + "-" * 80)
    print("CPU USAGE ANALYSIS (nfsd process)")
    print("-" * 80)
    print(f"Baseline:    {baseline_cpu_stats['median']:.2f}% ± {baseline_cpu_stats['ci_margin']:.2f}% (95% CI)")
    print(f"             {baseline_cpu_stats['std']:.2f}% std dev, range [{baseline_cpu_stats['min']:.2f}-{baseline_cpu_stats['max']:.2f}]%")
    if baseline_cpu_outliers:
        print(f"             ⚠ Outliers in runs: {baseline_cpu_outliers}")
    
    print(f"Detector:    {detector_cpu_nfsd_stats['median']:.2f}% ± {detector_cpu_nfsd_stats['ci_margin']:.2f}% (95% CI)")
    print(f"             {detector_cpu_nfsd_stats['std']:.2f}% std dev, range [{detector_cpu_nfsd_stats['min']:.2f}-{detector_cpu_nfsd_stats['max']:.2f}]%")
    if detector_cpu_outliers:
        print(f"             ⚠ Outliers in runs: {detector_cpu_outliers}")
    
    print(f"\nnfsd Overhead: {cpu_nfsd_overhead_pct:+.2f}%  (p={cpu_ttest['p_value']:.4f}, "
          f"effect size={cpu_ttest['effect_size']:.3f})")
    if cpu_ttest['is_significant']:
        print(f"               ✓ STATISTICALLY SIGNIFICANT (p < 0.05)")
    else:
        print(f"               ✗ NOT statistically significant")
    
    print("\n" + "-" * 80)
    print("DETECTOR MONITOR PROCESS")
    print("-" * 80)
    print(f"Avg CPU:     {detector_cpu_det_stats['median']:.2f}% ± {detector_cpu_det_stats['ci_margin']:.2f}% (95% CI)")
    print(f"             {detector_cpu_det_stats['std']:.2f}% std dev, range [{detector_cpu_det_stats['min']:.2f}-{detector_cpu_det_stats['max']:.2f}]%")
    print(f"Normalized:  {normalized_overhead:.3f}% of total CPU budget (8 cores)")
    
    # ========================================================================
    # IOPS & LATENCY ANALYSIS
    # ========================================================================
    baseline_iops = [r['fio_iops'] for r in baseline if r['fio_iops'] > 0]
    detector_iops = [r['fio_iops'] for r in detector if r['fio_iops'] > 0]
    
    if baseline_iops and detector_iops:
        baseline_iops_stats = calculate_statistics(baseline_iops)
        detector_iops_stats = calculate_statistics(detector_iops)
        iops_reduction = ((baseline_iops_stats['median'] - detector_iops_stats['median'])
                         / baseline_iops_stats['median'] * 100)
        iops_ttest = perform_t_test(baseline_iops, detector_iops)
        
        print("\n" + "-" * 80)
        print("IOPS (I/O Operations Per Second)")
        print("-" * 80)
        print(f"Baseline:    {baseline_iops_stats['median']:.0f} ± {baseline_iops_stats['ci_margin']:.0f} IOPS (95% CI)")
        print(f"Detector:    {detector_iops_stats['median']:.0f} ± {detector_iops_stats['ci_margin']:.0f} IOPS (95% CI)")
        print(f"Reduction:   {iops_reduction:+.2f}%  (p={iops_ttest['p_value']:.4f})")
        if iops_ttest['is_significant']:
            print(f"             ✓ STATISTICALLY SIGNIFICANT")
    
    baseline_lat = [r['fio_lat_ns'] for r in baseline if r['fio_lat_ns'] > 0]
    detector_lat = [r['fio_lat_ns'] for r in detector if r['fio_lat_ns'] > 0]
    
    if baseline_lat and detector_lat:
        baseline_lat_stats = calculate_statistics(baseline_lat)
        detector_lat_stats = calculate_statistics(detector_lat)
        lat_increase = ((detector_lat_stats['median'] - baseline_lat_stats['median'])
                       / baseline_lat_stats['median'] * 100)
        lat_ttest = perform_t_test(baseline_lat, detector_lat)
        
        print("\n" + "-" * 80)
        print("LATENCY (nanoseconds)")
        print("-" * 80)
        print(f"Baseline:    {baseline_lat_stats['median']:.0f} ± {baseline_lat_stats['ci_margin']:.0f} ns (95% CI)")
        print(f"Detector:    {detector_lat_stats['median']:.0f} ± {detector_lat_stats['ci_margin']:.0f} ns (95% CI)")
        print(f"Increase:    {lat_increase:+.2f}%  (p={lat_ttest['p_value']:.4f})")
        if lat_ttest['is_significant']:
            print(f"             ✓ STATISTICALLY SIGNIFICANT")
    
    # ========================================================================
    # DETAILED PER-RUN BREAKDOWN
    # ========================================================================
    if detailed:
        print("\n" + "-" * 80)
        print("DETAILED PER-RUN RESULTS")
        print("-" * 80)
        print("\nBaseline:")
        print(f"{'Run':<5} {'Runtime(s)':<12} {'BW(KB/s)':<12} {'IOPS':<12} {'nfsd CPU%':<10}")
        for r in baseline:
            iops_str = f"{r['fio_iops']:.0f}" if r['fio_iops'] > 0 else "N/A"
            print(f"{r['run_num']:<5} {r['fio_runtime_sec']:<12.3f} "
                  f"{r['fio_bw_kbps']:<12.0f} {iops_str:<12} {r['nfsd_avg_cpu_pct']:<10.2f}")
        
        print("\nDetector:")
        print(f"{'Run':<5} {'Runtime(s)':<12} {'BW(KB/s)':<12} {'IOPS':<12} "
              f"{'nfsd CPU%':<10} {'det CPU%':<10}")
        for r in detector:
            iops_str = f"{r['fio_iops']:.0f}" if r['fio_iops'] > 0 else "N/A"
            print(f"{r['run_num']:<5} {r['fio_runtime_sec']:<12.3f} "
                  f"{r['fio_bw_kbps']:<12.0f} {iops_str:<12} "
                  f"{r['nfsd_avg_cpu_pct']:<10.2f} {r['detector_avg_cpu_pct']:<10.2f}")
    
    # ========================================================================
    # SUMMARY
    # ========================================================================
    print("\n" + "=" * 80)
    print("SUMMARY")
    print("=" * 80)
    print(f"\nRuntime overhead:         {time_overhead_pct:+.2f}% "
          f"{'(significant)' if runtime_ttest['is_significant'] else '(not significant)'}")
    print(f"Bandwidth reduction:      {bw_overhead_pct:+.2f}% "
          f"{'(significant)' if bw_ttest['is_significant'] else '(not significant)'}")
    print(f"nfsd CPU overhead:        {cpu_nfsd_overhead_pct:+.2f}% "
          f"{'(significant)' if cpu_ttest['is_significant'] else '(not significant)'}")
    print(f"Detector CPU usage:       {detector_cpu_det_stats['median']:.2f}% "
          f"({normalized_overhead:.3f}% of budget)")
    print("\n" + "=" * 80 + "\n")



def export_plots(results: List[Dict], output_dir: str) -> None:
    """Generate matplotlib plots for visual analysis."""
    try:
        import matplotlib.pyplot as plt
        import numpy as np
    except ImportError:
        print("Error: matplotlib not installed. Install with: pip install matplotlib")
        return
    
    baseline = [r for r in results if r['mode'] == 'baseline']
    detector = [r for r in results if r['mode'] == 'detector']
    
    os.makedirs(output_dir, exist_ok=True)
    
    # Runtime comparison
    fig, ax = plt.subplots(figsize=(10, 6))
    
    baseline_times = [r['fio_runtime_sec'] for r in baseline]
    detector_times = [r['fio_runtime_sec'] for r in detector]
    
    ax.boxplot([baseline_times, detector_times], labels=['Baseline', 'Detector'])
    ax.set_ylabel('Runtime (seconds)')
    ax.set_title('FIO Runtime: Baseline vs. Detector')
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'runtime_comparison.png'), dpi=300)
    print(f"Saved: {output_dir}/runtime_comparison.png")
    
    # CPU comparison
    fig, ax = plt.subplots(figsize=(10, 6))
    
    baseline_cpu = [r['nfsd_avg_cpu_pct'] for r in baseline]
    detector_cpu_nfsd = [r['nfsd_avg_cpu_pct'] for r in detector]
    detector_cpu_det = [r['detector_avg_cpu_pct'] for r in detector]
    
    x = np.arange(3)
    width = 0.6
    
    medians = [
        statistics.median(baseline_cpu),
        statistics.median(detector_cpu_nfsd),
        statistics.median(detector_cpu_det)
    ]
    
    ax.bar(x, medians, width, color=['blue', 'orange', 'green'])
    ax.set_xticks(x)
    ax.set_xticklabels(['Baseline\nnfsd', 'Detector\nnfsd', 'Detector\nmonitor'])
    ax.set_ylabel('CPU Usage (%)')
    ax.set_title('CPU Usage Comparison')
    ax.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'cpu_comparison.png'), dpi=300)
    print(f"Saved: {output_dir}/cpu_comparison.png")
    
    plt.close('all')


def export_to_csv(results: List[Dict], output_path: str) -> None:
    """Export parsed results to CSV file."""
    if not results:
        print("Error: No results to export")
        return
    
    # Define CSV columns
    fieldnames = [
        'mode', 'run_num', 'fio_runtime_sec', 'fio_bw_kbps',
        'nfsd_avg_cpu_pct', 'detector_avg_cpu_pct',
        'nfsd_samples', 'detector_samples',
        'fio_iops', 'fio_lat_ns'
    ]
    
    with open(output_path, 'w', newline='') as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        
        for r in results:
            # Write only the fields we want
            row = {k: r.get(k, 0) for k in fieldnames}
            writer.writerow(row)
    
    print(f"\n✓ Exported results to: {output_path}")
    print(f"  {len(results)} runs written to CSV")



# ============================================================================
# MAIN
# ============================================================================

def main():
    if len(sys.argv) < 2 or '--help' in sys.argv:
        print(__doc__)
        sys.exit(0)
    
    input_path = sys.argv[1]
    
    if not os.path.exists(input_path):
        print(f"Error: Path not found: {input_path}")
        sys.exit(1)
    
    # Parse command line options
    reparse = '--reparse' in sys.argv
    detailed = '--detailed' in sys.argv
    export_plots_flag = '--export-plots' in sys.argv
    
    # Check for --export-csv option
    export_csv_path = None
    if '--export-csv' in sys.argv:
        try:
            idx = sys.argv.index('--export-csv')
            if idx + 1 < len(sys.argv):
                export_csv_path = sys.argv[idx + 1]
            else:
                print("Error: --export-csv requires a filename argument")
                sys.exit(1)
        except ValueError:
            pass
    
    # Determine if input is a directory or CSV file
    is_directory = os.path.isdir(input_path)
    
    # Load data
    if is_directory:
        print(f"Input is a directory, parsing raw data...")
        results = reparse_raw_data(input_path)
        
        # Auto-generate CSV filename if not specified
        if export_csv_path is None:
            dir_basename = os.path.basename(input_path.rstrip('/'))
            # Remove 'raw_data_' prefix if present
            if dir_basename.startswith('raw_data_'):
                csv_basename = dir_basename[9:] + '_reparsed.csv'
            else:
                csv_basename = dir_basename + '_reparsed.csv'
            export_csv_path = os.path.join(os.path.dirname(input_path) or '.', csv_basename)
        
        # Export to CSV
        export_to_csv(results, export_csv_path)
        print("")
        
    elif reparse:
        print("Re-parsing raw data for full reproducibility...")
        results = reparse_raw_data(input_path)
    else:
        results = load_csv_results(input_path)
    
    # Analyze
    analyze_results(results, detailed=detailed)
    
    # Export plots if requested
    if export_plots_flag:
        if is_directory:
            output_dir = os.path.join(input_path, 'plots')
        else:
            output_dir = os.path.join(os.path.dirname(input_path), 'plots')
        export_plots(results, output_dir)


if __name__ == '__main__':
    main()
```
